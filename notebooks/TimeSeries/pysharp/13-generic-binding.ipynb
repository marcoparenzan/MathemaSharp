# Binding CLR generico esplicito
Sintassi method[ClrType](arguments). Nessuna inferenza automatica. I parametri byref/Span non sono invocabili con MethodInfo.Invoke e sono esclusi dal binder. Il tipo CLR Decimal128 è diverso dalla funzione costruttrice decimal128.

```jupyternet
import clr
from mathema import decimal128
from System.Numerics import Decimal128
clr.AddReference("D:/Dev/BuildArtifacts/MathemaSharp/bin/MathemaSharp.LinearAlgebra/release/MathemaSharp.LinearAlgebra.dll")
from MathemaSharp.LinearAlgebra import MatrixAlgebra
matrix = MatrixAlgebra.Identity[Decimal128](2)
delta = decimal128("1e-25")
matrix[0, 0] = decimal128(1) + delta
det = MatrixAlgebra.Determinant2x2[Decimal128](matrix)
assert det - 1 == delta
assert matrix[1, 1] == decimal128(1)
print("generic_decimal128_determinant: OK")
transposed = MatrixAlgebra.Transpose[Decimal128](matrix)
assert transposed[0, 0] - 1 == delta
print("generic_matrix_copy: OK")
```
