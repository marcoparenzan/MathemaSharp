# PySharp ↔ MathemaSharp : Decimal128 e complessi
Facciata esplicita Decimal128Interop. I valori sono oggetti CLR: non diventano float Python. Creare valori da stringhe; un literal Python float perderebbe precisione prima della chiamata. Non prova il binding generico o il supporto Decimal di numpy.
Output eseguiti con kernel PySharp su runtime .NET 11, assembly Core Release locale.

```jupytersharp
import clr
clr.AddReference("D:/Dev/BuildArtifacts/MathemaSharp/bin/MathemaSharp.Core/release/MathemaSharp.Core.dll")
from MathemaSharp.Core.Interop import Decimal128Interop as D
one = D.Parse("1")
delta = D.Parse("1e-25")
value = D.Add(one, delta)
difference = D.Subtract(value, one)
assert D.Format(difference) == D.Format(delta)
print("decimal128_roundtrip: OK")
print("increment: " + D.Format(difference))
```

```jupytersharp
values = D.Copy([one, delta])
total = D.Sum(values)
assert D.Format(D.Subtract(total, one)) == D.Format(delta)
total_from_list = D.Sum([one, delta])
assert D.Format(total_from_list) == D.Format(total)
print("typed_array_and_list: OK")
```

```jupytersharp
zero = D.Parse("0")
z = D.Complex(one, delta)
identity = D.Complex(one, zero)
product = D.HermitianDot([z], [identity])
assert D.Equal(D.Real(product), one)
assert D.Equal(D.Imaginary(product), D.Subtract(zero, delta))
print("complex128_hermitian: OK")
```

## Limiti
Le operazioni sono chiamate esplicitamente nella facciata, non con operatori Python sugli oggetti CLR. La verifica riguarda Decimal128 e Complex<Decimal128>. Decimal32/64, inference dei metodi generici, numpy typed e pacchetti NuGet aggiornati sono attività successive.
