# PySharp ↔ MathemaSharp : Decimal128 and complex numbers
Explicit Decimal128Interop facade. The values are CLR objects: they do not become Python floats. Create values from strings; a Python float literal would lose precision before the call. It does not prove generic binding or numpy's Decimal support.
Outputs executed with the PySharp kernel on the .NET 11 runtime, local Release Core assembly.

```jupytersharp
import clr
clr.AddReference("D:/Dev/BuildArtifacts/MathemaSharp/bin/MathemaSharp.Core/release/MathemaSharp.Core.dll")
from MathemaSharp.Core.Interop import Decimal128Interop as D
one = D.Parse("1")
delta = D.Parse("1e-25")
value = D.Add(one, delta)
difference = D.Subtract(value, one)
assert D.Format(difference) == D.Format(delta)
print("decimal128_roundtrip: OK")
print("increment: " + D.Format(difference))
```

```jupytersharp
values = D.Copy([one, delta])
total = D.Sum(values)
assert D.Format(D.Subtract(total, one)) == D.Format(delta)
total_from_list = D.Sum([one, delta])
assert D.Format(total_from_list) == D.Format(total)
print("typed_array_and_list: OK")
```

```jupytersharp
zero = D.Parse("0")
z = D.Complex(one, delta)
identity = D.Complex(one, zero)
product = D.HermitianDot([z], [identity])
assert D.Equal(D.Real(product), one)
assert D.Equal(D.Imaginary(product), D.Subtract(zero, delta))
print("complex128_hermitian: OK")
```

## Limits
The operations are called explicitly on the facade, not with Python operators on the CLR objects. The check covers Decimal128 and Complex<Decimal128>. Decimal32/64, inference of generic methods, typed numpy and updated NuGet packages are later work.
