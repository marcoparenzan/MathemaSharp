# Modulo mathema e operatori CLR
Costruttori importabili dal runtime PySharp; nessun helper locale né facciata aritmetica. Richiede .NET 11 per i tipi IEEE Decimal. Accetta stringhe, interi e valori CLR dello stesso tipo; float impliciti e conversioni fra dtype diversi rifiutati. Arrotondamento/range restano quelli del tipo scelto.

```jupyternet
from mathema import decimal32, decimal64, decimal128
for make in [decimal32, decimal64, decimal128]:
    a = make(2)
    b = make(3)
    assert a + b == make(5)
    assert b - a == make(1)
    assert a * b == make(6)
    assert b / a == make("1.5")
    assert -a == make(-2)
    assert +a == a
    assert a < b and a <= b and b > a and b >= a
    assert a != b
    assert a + 1 == make(3) and 1 + a == make(3)
    print("decimal_operators: OK")
x = decimal128("1.0000000000000000000000001")
y = decimal128("1e-25")
assert x - 1 == y
assert -y + y == decimal128(0)
print("decimal128_native_increment: OK")
```
