# 01 — Generic Math

## Usage e contratti
Sum e Horner restituiscono zero per input vuoto. SumChecked usa aritmetica checked, ma i formati floating-point possono produrre infinito invece di OverflowException. Dot richiede lunghezze uguali; sui complessi e' bilineare, non hermitiano. Mean richiede input non vuoto e IFloatingPoint; non e' disponibile sugli interi. Gcd e' limitato a IBinaryInteger e Abs(MinValue) puo' traboccare. Factorial rifiuta n negativo e restituisce BigInteger.
[Reference completa](12-api-reference.ipynb), [verifiche sui sei formati](11-consolidation.ipynb).
Kernel **JupyterNet** (C#) su runtime **.NET 11**. Prerequisito: `dotnet build src/MathemaSharp.slnx -c Release`. Da terminale: `scripts/run-notebooks.ps1` (imposta il roll-forward al runtime 11); in VS Code serve `DOTNET_ROLL_FORWARD=LatestMajor` e, finché .NET 11 è RC, `DOTNET_ROLL_FORWARD_TO_PRERELEASE=1`.

`INumberBase<T>` basta per somme e prodotti, anche Complex e `Complex<T>` di .NET 11. `INumber<T>` aggiunge ordinamento; `IFloatingPoint<T>` evita una media con divisione intera.

```jupytersharp
#load "../shared/MathemaSharp.csx"
using System.Numerics;
using MathemaSharp.Core;
Console.WriteLine(NumericAlgorithms.Sum<int>(new int[] {1,2,3}));
Console.WriteLine(NumericAlgorithms.Mean<decimal>(new decimal[] {1,2}));
Console.WriteLine(NumericAlgorithms.Sum<Complex>(new Complex[] {new(1,2),new(3,4)}));
```

```jupytersharp
Console.WriteLine(NumericAlgorithms.Horner<double>(new double[] {1,2,1}, 3));
Console.WriteLine(byte.CreateSaturating(300));
Console.WriteLine(byte.CreateTruncating(300));
try { Console.WriteLine(byte.CreateChecked(300)); } catch (OverflowException) { Console.WriteLine("Overflow"); }
Console.WriteLine(NumericAlgorithms.Factorial(50));
Console.WriteLine(NumericAlgorithms.Gcd(48,18));
```

## Esperimenti
Sostituire double con decimal o BigInteger nel polinomio. Provare `SumChecked<int>` con MaxValue e 1. Provare `Mean<int>`: il compilatore lo rifiuta deliberatamente. Un constraint non garantisce associatività numerica né vettorizzazione.
