# 01 — Generic Math

## Usage and contracts
Sum and Horner return zero for empty input. SumChecked uses checked arithmetic, but floating-point formats can produce infinity instead of OverflowException. Dot requires equal lengths; on complex numbers it is bilinear, not Hermitian. Mean requires non-empty input and IFloatingPoint; it is not available on integers. Gcd is limited to IBinaryInteger and Abs(MinValue) can overflow. Factorial rejects negative n and returns BigInteger.
[Full reference](12-api-reference.ipynb), [checks on the six formats](11-consolidation.ipynb).
**JupyterSharp** kernel (C#) on the **.NET 11** runtime. Prerequisite: `dotnet build src/MathemaSharp.slnx -c Release`. From a terminal: `scripts/run-notebooks.ps1` (sets the roll-forward to runtime 11); in VS Code you need `DOTNET_ROLL_FORWARD=LatestMajor` and, as long as .NET 11 is RC, `DOTNET_ROLL_FORWARD_TO_PRERELEASE=1`.

`INumberBase<T>` is enough for sums and products, including Complex and .NET 11's `Complex<T>`. `INumber<T>` adds ordering; `IFloatingPoint<T>` avoids a mean computed with integer division.

```jupytersharp
#load "../shared/MathemaSharp.csx"
using System.Numerics;
using MathemaSharp.Core;
Console.WriteLine(NumericAlgorithms.Sum<int>(new int[] {1,2,3}));
Console.WriteLine(NumericAlgorithms.Mean<decimal>(new decimal[] {1,2}));
Console.WriteLine(NumericAlgorithms.Sum<Complex>(new Complex[] {new(1,2),new(3,4)}));
```

```jupytersharp
Console.WriteLine(NumericAlgorithms.Horner<double>(new double[] {1,2,1}, 3));
Console.WriteLine(byte.CreateSaturating(300));
Console.WriteLine(byte.CreateTruncating(300));
try { Console.WriteLine(byte.CreateChecked(300)); } catch (OverflowException) { Console.WriteLine("Overflow"); }
Console.WriteLine(NumericAlgorithms.Factorial(50));
Console.WriteLine(NumericAlgorithms.Gcd(48,18));
```
