# MàthemaSharp.Core - consolidamento

## Usage

Esempi e verifiche in T sui sei formati. Prerequisiti: build Release e kernel JupyterNet eseguito sul runtime .NET 11. I confronti sono su casi piccoli: non certificano tutte le API o l'accuratezza generale. Nessuna dipendenza Charts o NDSharp.



```jupytersharp
#load "../shared/MathemaSharp.Numerics.csx"
using System.Numerics;
using MathemaSharp.Core;
void Verify<T>() where T : IFloatingPointIeee754<T>, IMinMaxValue<T>
{
    T N(int n) => T.CreateChecked(n);
    T[] values = [T.One, N(2), N(3)];
    if (NumericAlgorithms.Sum<T>(values) != N(6)) throw new Exception("Sum");
    if (NumericAlgorithms.Mean<T>(values) != N(2)) throw new Exception("Mean");
    if (Signals.VectorDot<T>(values, values) != N(14)) throw new Exception("Dot");
    if (Precision.NearlyEqual(T.One, N(2))) throw new Exception("Tolerance");
    var bins = Signals.Dft<T>([Complex<T>.One, Complex<T>.Zero]);
    foreach (var bin in bins) if (bin != Complex<T>.One) throw new Exception("DFT");
    Console.WriteLine($"{typeof(T).Name}: sum=6 mean=2 dot=14 dft=OK tolerance=OK");
}
Verify<Half>();
Verify<float>();
Verify<double>();
Verify<Decimal32>();
Verify<Decimal64>();
Verify<Decimal128>();

```

## Reference - API usate e famiglie correlate

| API | Constraint / limite |
|---|---|
| Sum, Dot, Horner | INumberBase; Dot complesso bilineare |
| Mean | IFloatingPoint |
| NearlyEqual | IEEE; default min(4096 epsilon, sqrt(epsilon)) |
| Dft | IEEE + MinMax; O(N^2), non FFT |
| VectorDot | IEEE; SIMD solo se supportata, altrimenti fallback |
