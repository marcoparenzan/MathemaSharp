# MàthemaSharp.Quantum - consolidation

## Usage

Examples and checks in T on the six formats. Prerequisites: Release build and a JupyterSharp kernel running on the .NET 11 runtime. The comparisons are on small cases: they do not certify all the APIs or general accuracy. No Charts or NDSharp dependency.



```jupytersharp
#load "../shared/MathemaSharp.Numerics.csx"
using System.Numerics;
using MathemaSharp.Quantum;
void Verify<T>() where T : IFloatingPointIeee754<T>, IMinMaxValue<T>
{
    T N(int n) => T.CreateChecked(n);
    var eps = T.BitIncrement(T.One)-T.One;
    var unit = QuantumAlgorithms.Normalize<T>([new(T.MaxValue,T.Zero),new(T.MaxValue,T.Zero)]);
    if (T.Abs(QuantumAlgorithms.Norm<T>(unit)-T.One) > N(64)*eps) throw new Exception("Normalize");
    var eigen = QuantumAlgorithms.CheckEigenstate<T>(Pauli<T>.X,unit,N(64)*eps);
    if (!eigen.IsEigenstate) throw new Exception("Eigenstate");
    var diagonal = QuantumAlgorithms.DiagonalizeHermitian2(Pauli<T>.Y);
    if (T.Abs(diagonal.LambdaPlus-T.One) > N(64)*eps) throw new Exception("Diagonalization");
    var evolution = QuantumAlgorithms.TimeEvolution(Pauli<T>.X,T.Pi/N(2));
    if (!evolution.IsUnitary(N(64)*eps)) throw new Exception("MatrixExp");
    Console.WriteLine($"{typeof(T).Name}: normalization=OK eigenstate=OK diagonalization=OK evolution=OK");
}
Verify<Half>();
Verify<float>();
Verify<double>();
Verify<Decimal32>();
Verify<Decimal64>();
Verify<Decimal128>();

```

## Reference - APIs used and related families

| API | Constraint / limit |
|---|---|
| Norm / Normalize | IEEE + MinMax; Hypot and scaling |
| CheckEigenstate | Residual relative to the Frobenius norm of the operator |
| DiagonalizeHermitian2 | 2x2 only; limits for unbalanced spectra |
| MatrixExp | Taylor (24 terms by default), not Pade |
| TimeEvolution | finite t, finite positive hbar |
