# NDSharp contour integration through a real PySharp kernel
Source-build smoke: nonuniform physical coordinates, line contours delegated to PlotSharp, filled bands retained in NDSharp. Runner: dotnet run --file ops/RunNDSharpContourNotebook.cs, after dotnet run --file ops/BuildPySharpKernel.cs -- --source-plots. This is JupyterNet plugin execution, not native JupyterSharp integration or an accuracy oracle. PNG outputs remain embedded in this notebook; no review screenshot/image files are produced.

```jupytersharp
import numpy as np
import matplotlib.pyplot as plt
xs = np.array([0.0, 1.0, 10.0])
ys = np.array([0.0, 1.0, 4.0])
z = np.array([[0.0, 1.0, 2.0], [0.0, 1.0, 2.0], [0.0, 1.0, 2.0]])
print('NONUNIFORM_CONTOUR_INPUT_READY')
```

```jupytersharp
fig, ax = plt.subplots(figsize=(3.2, 2.4), dpi=100)
ax.contour(xs, ys, z, levels=[0.5, 1.5], colors=['red'])
ax.set_xlim(0.0, 10.0)
ax.set_ylim(0.0, 4.0)
ax.set_title('Line contours')
plt.show()
print('NDSHARP_LINE_CONTOUR_KERNEL_OK')
```

```jupytersharp
fig, ax = plt.subplots(figsize=(3.2, 2.4), dpi=100)
ax.contourf(xs, ys, z, levels=[0.0, 1.0, 2.0], colors=['blue', 'red'])
ax.set_xlim(0.0, 10.0)
ax.set_ylim(0.0, 4.0)
ax.set_title('Filled bands')
plt.show()
print('NDSHARP_FILLED_CONTOUR_KERNEL_OK')
```
