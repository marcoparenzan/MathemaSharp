# MàthemaSharp → PlotSharp → JupyterSharp
Usage: a signal and its FFT rendered by PlotSharp, executed by the actual JupyterSharp CLI from NuGet. Run `dotnet run --file ops/RunEcosystemPilot.cs` first to restore packages.
Reference: FFT forward is unnormalized; plot coordinates are explicitly double. This example does not certify Decimal plotting or PySharp kernel integration.

```jupytersharp
#r "D:/Dev/BuildArtifacts/net11-pack/consumer-cache/mathemasharp.core/0.2.1-net11-preview.1/lib/net11.0/MathemaSharp.Core.dll"
#r "D:/Dev/BuildArtifacts/net11-pack/consumer-cache/plotsharp/0.1.1-net11-preview.2/lib/net11.0/PlotSharp.dll"
using System.Numerics;
using MathemaSharp.Core;
using PlotSharp;
using JupyterSharp.Kernels.CSharp;
Console.WriteLine("PACKAGED_ECOSYSTEM_READY");
```

```jupytersharp
int n = 64;
double[] time = Enumerable.Range(0, n).Select(i => (double)i / n).ToArray();
double[] signal = time.Select(t => Math.Sin(2 * Math.PI * 5 * t)).ToArray();
Complex<double>[] input = signal.Select(y => new Complex<double>(y, 0)).ToArray();
var spectrum = Signals.Fft<double>(input);
double[] magnitude = spectrum.Take(n / 2 + 1).Select(z => double.Hypot(z.Real, z.Imaginary) / n).ToArray();
int peak = Enumerable.Range(0, magnitude.Length).MaxBy(i => magnitude[i]);
if (peak != 5 || Math.Abs(magnitude[5] - 0.5) > 1e-12) throw new Exception("FFT peak incorrect");
Console.WriteLine("FFT_PEAK_5_VERIFIED");
```

```jupytersharp
var signalSvg = Charts.LineChart([new Charts.Series("Signal", time, signal, Charts.Series1)], new Charts.LineChartOptions { Title = "Signal: 5 cycles", XLabel = "Time", YLabel = "Amplitude" });
var spectrumSvg = Charts.LineChart([new Charts.Series("Magnitude", Enumerable.Range(0, magnitude.Length).Select(i => (double)i).ToArray(), magnitude, Charts.Series2, Line: false, Stems: true)], new Charts.LineChartOptions { Title = "FFT spectrum", XLabel = "Bin", YLabel = "Magnitude / N" });
if (!signalSvg.Contains("<svg") || !spectrumSvg.Contains("<svg")) throw new Exception("No SVG");
Display.Html(signalSvg);
Display.Html(spectrumSvg);
Console.WriteLine("PLOTSHARP_TWO_SVG_VERIFIED");
```
