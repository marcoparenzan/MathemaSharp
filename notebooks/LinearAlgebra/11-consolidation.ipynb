# MàthemaSharp.LinearAlgebra - consolidation

## Usage

Examples and checks in T on the six formats. Prerequisites: Release build and a JupyterSharp kernel running on the .NET 11 runtime. The comparisons are on small cases: they do not certify all the APIs or general accuracy. No Charts or NDSharp dependency.



```jupytersharp
#load "../shared/MathemaSharp.Numerics.csx"
using System.Numerics;
using MathemaSharp.LinearAlgebra;
void Verify<T>() where T : IFloatingPointIeee754<T>, IMinMaxValue<T>
{
    T N(int n) => T.CreateChecked(n);
    var eps = T.BitIncrement(T.One) - T.One;
    T[,] a = {{T.One,T.Zero},{T.Zero,T.One},{T.One,T.One}};
    var beta = MatrixAlgebra.LeastSquares<T>(a, [N(2),N(3),N(5)]);
    if (T.Abs(beta[0]-N(2)) > N(64)*eps || T.Abs(beta[1]-N(3)) > N(64)*eps) throw new Exception("QR");
    var lu = LuFactorization<T>.Create(new T[,] {{T.Zero,T.One},{T.One,T.One}});
    var x = lu.Solve([N(3),N(5)]);
    if (T.Abs(x[0]-N(2)) > N(64)*eps || T.Abs(x[1]-N(3)) > N(64)*eps) throw new Exception("LU");
    Console.WriteLine($"{typeof(T).Name}: QR=OK LU=OK pivoting=OK");
}
Verify<Half>();
Verify<float>();
Verify<double>();
Verify<Decimal32>();
Verify<Decimal64>();
Verify<Decimal128>();

```

## Reference - APIs used and related families

| API | Constraint / limit |
|---|---|
| Multiply, Identity | INumber |
| LeastSquares | IEEE; normalised QR, does not estimate conditioning |
| LuFactorization | IEEE; partial pivoting, reusable |
| NormL2/FrobeniusNorm | IEEE, Hypot |
| Inverse2x2 | IEEE; rejects near-singularity |
